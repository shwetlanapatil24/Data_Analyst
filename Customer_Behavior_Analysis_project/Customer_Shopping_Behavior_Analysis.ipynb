```python
#How should be replace the null values?
#probably we will replace it with mean or median
#generally we should choose median over mean because mean can get affected by outliers and would skew the distribution of our data.
#and why median ? because median is robust to outliers


df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns = {'purchase_amount_(usd)': 'purchase_amount'})   #we replaced the name of purchase_amount_(usd) with purchase_amount
```

```python
df.columns
```

```python
# create a new column age_group
#qcut splits the ages into 4 equal size groups based on data distribution and asigns the lables we defind!

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
#In pandas, the number of brackets determines whether you are selecting one column or multiple columns.
#Why double brackets [[ ]] are required [['age','age_group']] is a list of column names. pandas expects a list when you want multiple columns.
#means: “Give me the columns age and age_group.” 
#What happens with a single bracket ? Python interprets this as a tuple, not a list, pandas then tries to find one column with the name,
#which does not exist, so you get an error like KeyError: ('age', 'age_group')


df[['age','age_group']].head()
```

```python
# create new column purchase_frequency_days
#we'll convert the texual frequencies into number

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
